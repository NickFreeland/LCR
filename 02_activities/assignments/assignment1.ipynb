```python
np.random.seed(1)
from sklearn.model_selection import train_test_split
X = wine_df.drop(columns=['class'])
y = wine_df['class']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=1)
```

Question 3:
Model initialization and cross-validation
We are finally set to fit the KNN model.

Perform a grid search to tune the n_neighbors hyperparameter using 10-fold cross-validation. Follow these steps:

1. Initialize the KNN classifier using KNeighborsClassifier().
2. Define a parameter grid for n_neighbors ranging from 1 to 50.
3. Implement a grid search using GridSearchCV with 10-fold cross-validation to find the optimal number of neighbors.
4. After fitting the model on the training data, identify and return the best value for n_neighbors based on the grid search results.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.model_selection import GridSearchCV

#1 Initialize the KNN classifier
knn = KNeighborsClassifier()

#2 Define the parameter grid for 'n_neighbors'
param_grid = {'n_neighbors': list(range(1, 50))}

#3 Set up GridSearchCV with 10-fold cross-validation
grid_search = GridSearchCV(estimator=knn, param_grid=param_grid, cv=10)

#4 Fit the GridSearchCV to the training data
grid_search.fit(X_train, y_train)

#5 Get the best 'n_neighbors' value
best_k = grid_search.best_params_['n_neighbors']
print(f"The best value of k is: {best_k}")
```

Question 4:
Model evaluation
Using the best value for n_neighbors, fit a KNN model on the training data and evaluate its performance on the test set using accuracy_score.

```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

knn_best = KNeighborsClassifier(n_neighbors=best_k)
knn_best.fit(X_train, y_train)
y_pred = knn_best.predict(X_test)
accuracy = accuracy_score(y_test, y_pred)
print("Test Accuracy:", accuracy)
```
